```python

import math

# a. luas persegi panjang
def luas_pp(panjang, lebar):
  return panjang * lebar

a = luas_pp(10,6)
print('luas persegi panjangnya: ', a)


#b. volume balok
def volume_balok(panjang, lebar, tinggi):
  return panjang * lebar * tinggi

b = volume_balok(10,8,6)
print('volume balok: ', b)


#c. luas lingkaran
def luas_lingkaran(r):
  jarijari = r ** 2
  return math.pi * jarijari

c = luas_lingkaran(7)
print('luas lingkaran: ', c)


#d. volume bola
def volume_bola(r):
  jarijari = r ** 3
  return (4/3) * math.pi * jarijari

d = volume_bola(7)
print('volume bola: ', d)
```
